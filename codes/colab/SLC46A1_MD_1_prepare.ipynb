This writes `md/inputs/build_spec.json`, which the three part 2 notebooks and part 3
all read. Nothing downstream types a position, a residue or a count. The spec names
the 26 substituted positions of each arm, the residue each arm must carry at every
one of them, the 67 positions of the pharmacophore set, the 72 with 306 disulfide the build
must carry, the formal charge of each arm, and the 2 positions where the sequence
likelihood and the structure prediction disagree.

The three arms are the wild-type chicken transporter, the 20 humanising substitutions
alone, and the construct, which is those 20 with 6 rescuers. The middle arm is there
because without it nothing in the simulations separates the rescuers helping from the
humanisation being tolerated without them.

After this, run the CHARMM-GUI builds with `md/inputs/CHARMM-GUI-checklist.txt`, put
each `gromacs/` folder on Drive where the checklist says, and open the three part 2
notebooks. They are independent and can run at the same time.
